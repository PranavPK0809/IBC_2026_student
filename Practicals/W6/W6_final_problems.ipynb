```python
import pickle

genetic_code = pickle.load(open("genetic_code.pickle", "rb"))
```

```python
test_mRNA = "AUGGAAUUCUCGCUCUGAAGGUAA"
```

```python
def get_amino_acids(mRNA):
    i = 0
    aa_sequence = []
    while (i + 3) < len(mRNA):
        codon = mRNA[i:(i + 3)]
        aa = genetic_code[codon]
        if aa == "Stop":
            break
        else:
            aa_sequence.append(aa)
        i = i + 4
    return "".join(aa_sequence)
```

```python
with open("Turkey_transcripts_15_coding.fasta", "r") as infile, open("Turkey_proteins_15.fasta", "w") as outfile:

    for line in infile:
        line = line.rstrip()

        if line[0] == ">":
            parts = line.split("gbkey=CDS")
            new_header = parts[0] + "gbkey=AA"
            outfile.write(new_header + "\n")

        else:
            mrna = ""

            for base in line:
                if base == "T":
                    mrna = mrna + "U"
                else:
                    mrna = mrna + base

            protein = get_amino_acids(mrna)
            outfile.write(protein + "\n")
```

```python
protein_file = open("Turkey_proteins_15.fasta", "r")

count = 0

for line in protein_file:
    line = line.rstrip()

    if line[0] == ">":
        count = count + 1

print(count)
```
